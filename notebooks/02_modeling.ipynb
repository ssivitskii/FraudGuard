# Reproduce the chronological benchmark
Run from this repository after installing requirements-benchmark.txt and downloading the original ULB CSV. No outputs are precomputed in this notebook.


```python
import json
import subprocess
import sys
from pathlib import Path

repo = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
subprocess.run([sys.executable, "-m", "scripts.benchmark", "--data", "data/raw/creditcard.csv", "--output", "reports/creditcard.json"], cwd=repo, check=True)
report = json.loads((repo / "reports/creditcard.json").read_text())
report["selected_model"], report["test"], report["limitations"]
```
